Al principio, el servicio no arrancaba y salía con un error `status=1/FAILURE`.  
Mirando los logs (`/var/log/mosquitto/mosquitto.log`), aparecía el típico:

> **Address already in use**

Básicamente, había un lío porque Mosquitto estaba intentando usar el puerto **1883** dos veces (IPv4 e IPv6), debido a varios archivos de configuración duplicados.

---
Para solucionarlo se hizo lo siguiente:

- Se cerraron procesos que se habían quedado colgados.
- Se limpió la carpeta de configuraciones extra.
- Se rehizo el archivo principal:

`/etc/mosquitto/mosquitto.conf`

Y se dejó configurado para:
- Escuchar solo en IPv4 (`0.0.0.0`)
- Permitir conexiones sin usuario/contraseña (`allow_anonymous true`)

```conf
listener 1883 0.0.0.0
allow_anonymous true

![Mosquitto funcionando](img.png)
